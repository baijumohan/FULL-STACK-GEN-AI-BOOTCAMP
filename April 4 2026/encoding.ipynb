```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
import numpy as np

# Create dummy data
data = {
    'Color': ['Red', 'Blue', 'Green', 'Red', 'Blue'],
    'Size': ['Small', 'Medium', 'Large', 'Medium', 'Small'],
    'Shape': ['Circle', 'Square', 'Triangle', 'Circle', 'Square']
}

df = pd.DataFrame(data)
print("Original Data:")
print(df)
```

```python
# Method 1: Using pandas get_dummies (simple and easy)
encoded_df_pandas = pd.get_dummies(df, columns=['Color', 'Size', 'Shape'])
print("\nMethod 1 - Pandas get_dummies (One Hot Encoded Data):")
print(encoded_df_pandas)
print(f"Shape: {encoded_df_pandas.shape}")
```

```python
# Method 2: Using sklearn OneHotEncoder (more powerful for pipelines)
encoder = OneHotEncoder(sparse_output=False)
encoded_array = encoder.fit_transform(df[['Color', 'Size', 'Shape']])

# Convert to DataFrame for better visualization
encoded_df_sklearn = pd.DataFrame(
    encoded_array, 
    columns=encoder.get_feature_names_out(['Color', 'Size', 'Shape'])
)
print("\nMethod 2 - Sklearn OneHotEncoder (One Hot Encoded Data):")
print(encoded_df_sklearn)
print(f"Shape: {encoded_df_sklearn.shape}")
```
